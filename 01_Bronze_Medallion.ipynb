# 01_Bronze_Medallion

Esta notebook obtiene un archivo CSV público a través de una API REST - mediante el método GET, guarda los datos sin procesar en una carpeta `bronze` y preserva el conjunto de datos tal como está para la primera capa de medallón. El conjunto de datos de origen es el CSV de muestra de Superstore desde un endpoint de GitHub público.

```python
import os
from pathlib import Path
import requests
import pandas as pd
```

### 1. Configuración de rutas y URLs

```python
BRONZE_DIR = Path('data/bronze')
BRONZE_DIR.mkdir(parents=True, exist_ok=True)
RAW_CSV_PATH = BRONZE_DIR / 'superstore_raw.csv'
DATA_URL = 'https://raw.githubusercontent.com/plotly/datasets/master/superstore.csv'
print('CSV guardado en:', RAW_CSV_PATH.resolve())
```

### 2. Descarga

```python
# Descargar el CSV desde el endpoint público via API REST y guardarlo localmente
response = requests.get(DATA_URL, timeout=20)
response.raise_for_status()
RAW_CSV_PATH.write_text(response.text, encoding='utf-8')
print('Downloaded', len(response.text.splitlines()), 'lines from the public REST endpoint.')
```

### 3. Almacenamiento

```python
# Cargar el CSV para verificar su contenido
df = pd.read_csv(RAW_CSV_PATH)
print('Raw dataset shape:', df.shape)
print(df.head(5).to_string(index=False))
```
